```python
# Decorator to note down the time

import time

def time_it(func):
    def wrapper(*args, **kwargs):
        start_t = time.time()
        result = func(*args, **kwargs)
        end_t = time.time()
        time_taken = end_t - start_t
        print(f"Time Taken : {time_taken}")
        return time_taken   # return timing
    return wrapper


@time_it
def python_add():
    a = list(range(1_000_000))
    b = list(range(1_000_000))
    c = [x + y for x, y in zip(a, b)]
    return c   # optional but good practice


time_taken = python_add()
print("Returned time:", time_taken)
```

```python
a = list(range(1000000))
b = list(range(1000000))

start_t = time.time()
c = [x + y for x,y in zip(a,b)]
end_t = time.time()
time_taken_p = end_t - start_t
print(f"Time Taken: {time_taken_p}")
```

```python
time_taken_p/time_taken
```
