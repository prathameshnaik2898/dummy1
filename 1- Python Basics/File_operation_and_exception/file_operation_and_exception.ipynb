```python
data_of_students = {}
```

```python
def populate_data(data):
    print(data)
    data = data.split(",")
    print(data)
    data_of_students[str(data[0])] = {}
    data_of_students[str(data[0])][str("name")] = data[1]
    data_of_students[str(data[0])][str("age")] = data[2]
```

```python
with open("test_dummy_1.txt", "r+") as f:
    content = f.readline()
    for c in content:
        populate_data(c)
```
