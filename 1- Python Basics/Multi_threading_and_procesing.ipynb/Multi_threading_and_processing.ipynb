# Without Threading

```python
import requests    # This library will help us to run URLs
import time        
```

```python
url = "https://raw.githubusercontent.com/Monalsingh/VideoBroadcaster/refs/heads/main/static/default-office-animated.png"
```

```python
def download_file(url, file_path ):      # file_path → where to save the file on your system
    try:
        response = requests.get(url)        # Sends a GET request to the URL, Downloads the content into response,response.content → raw bytes of file
        print(response) # Succes is 200 that means the response is successful
        with open(file_path, "wb") as file:         # w is write mode(over writes if file exists) and b is binary mode
            for chunk in response.iter_content(chunk_size = 8192): # Reads the download file piece by piece,Each chunk is 8192 bytes(8KB)
                if chunk:
                    file.write(chunk)
        print("File downloaded Successfully")
    except Exception as e:
        print(f"Error downloading file: {e}")
```

```python
download_file(url,"a.png")     # Saving it in a.png file in our local system from web 
```

```python
t1 = time.time()
download_file(url,"a.png")
download_file(url,"b.png")
download_file(url,"c.png")
t2 = time.time()
print(f"Time taken(seconds) : {t2-t1}")
```

```python
# downloading the files concurrently 

def download_file(process_name, url, file_path ):      # file_path → where to save the file on your system
    try:
        print(f"Download process name Started :{process_name}")
        response = requests.get(url)        # Sends a GET request to the URL, Downloads the content into response,response.content → raw bytes of file
        
        with open(file_path, "wb") as file:         # w is write mode(over writes if file exists) and b is binary mode
            for chunk in response.iter_content(chunk_size = 8192): # Reads the download file piece by piece,Each chunk is 8192 bytes(8KB)
                if chunk:
                    file.write(chunk)
        print("File downloaded Successfully")
    except Exception as e:
        print(f"Error downloading file: {e}")
    print(f"Process name completed: {process_name}")
```

```python
t1 = time.time()
download_file("Download without thread 1",url,"a.png")
download_file("Download without thread 2",url,"b.png")
download_file("Download without thread 3",url,"c.png")
t2 = time.time()
print(f"Time taken(seconds) : {t2-t1}")
```

# With Threading 

```python
import threading
```

```python
# Creating a thread (thread would have multiple tasks )
# We have 3 arguments because of 

t1= threading.Thread(target = download_file, args =("Download with thread 1", url,"a1.png"))   # downlod_file function has this parameter ("Download without thread 1",url,"a.png")

t2= threading.Thread(target = download_file, args =("Download with thread 2", url,"b1.png"))

t3= threading.Thread(target = download_file, args =("Download with thread 3", url,"c1.png")) 
```

# thread.start()

```python
t1_t = time.time()
t1.start()
t2.start()
t3.start()
print("Main Program done!!")
t2_t = time.time()
print(f"Time taken(seconds): {t2_t -t1_t}")
```

# thread.join()

#### Execution flow:

Main thread starts t1

Main thread starts t2

Main thread starts t3

Main thread waits for t1 to finish

Then waits for t2

Then waits for t3

ONLY AFTER ALL finish:


```python
t1_t = time.time()
t1.start()
t2.start()
t3.start()

t1.join()                                        
t2.join()   # join() blocks the main thread until the child thread completes, ensuring correct execution order and accurate timing.
t3.join()

print("Main Program done!!")
t2_t = time.time()
print(f"Time taken(seconds): {t2_t - t1_t}")
```

```python
# Can you control the running time of t1, t2 and t3 ? Answer is No.
# When we run t1.join(),t2.join(),t3.join() it means we want to complte the threads first and then print the last two statements
```
