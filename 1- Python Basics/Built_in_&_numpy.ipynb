# Using key = in both sort and sorted 

```python
words = ["apple", "kiwi", "banana"]

# No of characheters in words
# apple = 5
# kiwi = 4
# banana = 6 

# kiwi, apple, banana ---> Sorting by number of characters
```

```python
words.sort(key = len)
```

```python
words
```

```python
words = ["apple", "kiwi", "banana"]
print(sorted(words, key = len))
```

```python
# Sorting by last letter 

words = ["apple", "kiwi", "banana"]
print(sorted(words, key = lambda x: x[-1]))        # - 1 gives you last letter 
```

```python
def get_last_char(x):
    return x[-1]
```

```python
words = ["apple", "kiwi", "banana"]
# apple = e
# kiwi = i
# banana = a
print(sorted(words, key = get_last_char))
```

```python
words = ["apple", "kiwi", "banana"]
# apple = e
# kiwi = i
# banana = a
print(sorted(words, key = get_last_char, reverse = True))
```

# List Comprehension
### Its a short expressive way to generate a new list or filter a list

```python
# Problem Statement - Keep all the ones 

nums = [1,1,1,1,1,1,1,2,2,2,3,4,1,1,1,234,1,5,6]
```

```python
# Syntax : [expression for item in iterable if condition]

new_res = []
for i in nums:
    if i ==1:
        new_res.append(i)
print(new_res)
```

```python
new_res = [i for i in nums if i == 1]      # With list comprehension different for if else
```

```python
# Problem Statement : Square of numbers

new_res = []
for i in nums:
    new_res.append(i*i)
print(new_res)
```

```python
new_res = [i*i for i in nums]   # Diferent for for loop
```

```python
new_res
```

```python
# Problem statement - If number is even, then perform the square and output should be in list
nums = [1,2,3]
squares = []

for i in nums:
    if i % 2 == 0:
        square = i * i
        squares.append(square)
print(squares)
```

```python
# List Comprehension
nums = [1, 2, 3]
squares = [i*i for i in nums if i % 2 == 0]
print(squares)
```

# Dictionary Comprehension 

```python
# Dictionary Comprehension 
# Syntax {key_expression: value_expression for item in iterable if condition}

```

```python
names = ["Monal", "Gagan", "hello"]
scores = [10,100,-100]

result = {name.capitalize(): score + 100 for name, score in zip(names,scores)}
```

```python
print(result)
```
