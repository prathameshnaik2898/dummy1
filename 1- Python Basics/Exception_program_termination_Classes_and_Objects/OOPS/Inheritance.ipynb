```python
# Task is to allot manager and their team size 

class Employee:
    def __init__(self, name, senior_status):            # Self is address of the class itself.
        self.name = name                                    # self is used to communicate within the class
        self.seniority = senior_status
        print(f"Employee Name: {self.name}, Status: {self.seniority}")

class Manager(Employee):      # Manager is also employee so inheriting employee in Manger class
    def __init__(self, name, senior_status, team_size):  # Task is to run the constructor of Employee from Manager class
        super().__init__(name, senior_status)     # To call employee use the word super() - super means the class which we are inhereting   
        self.team_size = team_size                                  # So now super().__init__ has inheritance of Employee class

    def show(self):
        print(f"Manages {self.team_size} people")









```

```python
m = Manager("Monal", "senior", "100")   # If we run manager, Manager class will be created name,seniority, team_size will be updated as assigned
                                # Constructor will run in this constructor we are calling the name of the inherited class Employee, then we are passing 
                                # the name which we got senor status we got and then the whole flow will go into the Employee class everything will be assigned and then we will print.
                                # after that it will come down to Manager class and team size would be assigned and then print it 
```

```python
# Abstarct Base Class

class MacDonalds(ABC):

    @abstractmethod
    def mincapacity30(self):
        pass 

    @abstractmethod
    def miniparking10(self):
        pass 
    
    @abstractmethod
    def minivegNonvegSections(self):
        pass
```

```python
class Monal(MacDonalds):
    def mincapacity30(self):
        print("We hav seating capacity of 40.")
    def miniparking10(self):
        print("We have maximum parking capacity of 400")
    def minivegNonvegSections(self):
        print("We prepare veg and non-veg seperately")
    def newburger(self):
        print("Hey!!, this is a new burger.")
```

```python
mcd_new = Monal()
```

```python
mcd_new.newburger()
```
