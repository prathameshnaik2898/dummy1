```python
# Car 
# - Customer name
# - brand -->
# - color --> 
# - type of car --> Sedan, hatchback, sub

class Car():                  # Always start class with a Capital alphabet for defining class 
    
    def customer_name(self,name):     # Self indicates it is the address of Car class. First python will save functions address and then indicate to car class
        print(name)
        
    def brand(self,brand):
        print(brand)

    def color(self,color):
        print(color)

    def type_of_car(self,type_of_car):
        print(type_of_car)


    
```

```python
car_var = Car()    # Creating a varible to call Car class
```

```python
id(car_var)          # This is the id / address of the car_var
```

```python
car_var.customer_name("Monal")     # Storing/ Pulling customer_name inside the car Class by using car_var as we have assigned car_var to car class
```

```python
car_var_2 = Car()
car_var_2.customer_name("Hello") 
```

```python
class Mobile():                  # Always start class with a Capital alphabet for defining class 
    
    def __init__(self, manufacturer_name, os_version, customer_name):         # self --> It is the current class address of the class rest (manufacturer_name tp customer_name they are called as arguments)
        self.manufacturer_name = manufacturer_name             # a is a variale binded to the Class Mobile
        self.os_version = os_version
        self.customer_name = customer_name


    def get_customer_name(self):     # This is function (get_customer_name)
        print(self.customer_name)
      
```

```python
customer_1 = Mobile("Apple", "ios", "monal")
customer_2 = Mobile("Apple", "ios", "Akash")       # all of these are object names
customer_3 = Mobile("Apple", "ios", "Alok")        # When we create object from a class the first thing runs automatically is init
```

```python
customer_1.get_customer_name()            # get_customer_name is function
```

```python
customer_2.get_customer_name()
```
