# Try Except Block

```python
try :
    for i in range(5):
        a = int(input("Enter a number"))
        a = int(a)    # Here you have assigned integer as a type so it will only take numbers if you pass string it will crash 
        print(a)
        
except ValueError:
    print("Issue with the program")

print("Hello World")
```

```python
try :
    a = input("Enter a value: ")
    a = int(a)
    b = input("Enter a Value to convert")
    print(10/a)
    int(b)

except (ValueError,ZeroDivisionError):
    print("Issue with the program")

print("Hello World")
```

```python
try :                                      # This try block is used to handel exceptions
    a = input("Enter a value: ")
    a = int(a)
    b = input("Enter a Value to convert")
    print(10/a)
    int(b)

except Exception as e:                     # This Exception function will handel all error --> e is alias for Exception 
    print("Issue with the program:", e)                # Exception is a parent of all handeling error

print("Hello World")
```

```python
def take_age(age):
    try:
        age_int = int(age)
        if age_int < 0 or age_int > 100:
            raise ValueError("Age is not in range, Please try with a different input")
        else:
            print(age_int)

    except Exception as e:
        print(e)
    finally:
        print("Code Executed Successfully")
   
```

```python
take_age(-1)
```

```python
take_age(2)
```
