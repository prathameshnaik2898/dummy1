# More functions on Lists and Sets 


```python
a = [1,2,3,4,5,6,7,8,9,1,1,1,1,1,1]       # This is a list with duplicates
a = set(a)                                 # Removing duplicates using Sets
print(a)
```

```python
list(set(a))           # Converting sets back to list after duplicate removal
```

```python
# Extend 

a = [1,2,3,4]
b = ["Hello", "Hi"]
print(id(a))
a.extend(b)
print(id(a))
print(a)
```

```python
a = {1,2,3,4}
b = {"Hello", "Hi"}
print(id(a))
a.update(b)
print(id(a))
print(a)
```
