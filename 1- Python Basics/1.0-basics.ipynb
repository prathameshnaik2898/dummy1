```python
# Simple Calculator 

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

sum = num1 + num2
difference = num1 - num2 
product = num1 * num2 
quotient = num1 / num2 if num2 != 0 else "undefined"

print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```

```python
operation_name = "add"

if operation_name == "add":
    print("We are adding two numbers")
    print(2+3)
elif operation_name == "subtract":
    print("We are subtracting two numbers")
    print(5-2)  
else:
    print("We are neither adding nor subtracting")

print("after the control flow")
```

```python
operation_name = "add"
if operation_name == "add":
    print("We are adding two numbers")
    sum = 2 + 3 
    print(f"The sum of two numbers after adding is {sum}")
elif operation_name == "Subtract":
    print("We are subtracting two numbers")
elif operation_name == "multiply":
    print("We are multiplying two numbers")
elif operation_name == "divide":
    print("We are dividing two numbers")
else:
    print("No valid operation found")
```
