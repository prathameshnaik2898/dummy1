```python
# 3. repr - This method is used to show official string representation of the object, only used for dubugging. This is formal developer focused string 


class Person:
    def __init__(self,name):   
        self.name = name 
                                    
    def __repr__(self):                         # Str and Repr both looks same 
        return f"Person('{self.name}')"          # We use Str for Important information 
                                                    # repr we use for class invocation values 
    def __str__(self):
        return f"Person object has these values: {self.name}"
    
    def add(self, a, b):
        print(a + b)
```

```python
p_repr = Person("Monal")   # This is repr object
```

```python
print(p_repr)       # This is printing str using repr object 
```

```python
print(repr(p_repr))    # This show the class by back tracking 
```

```python
# 4. add

class NewAdd:
    def __init__(self, x, y):
        self.x = x 
        self.y = y
    
    def __repr__(self):
        return f"NewAdd({self.x}, {self.y})"
    
    def __add__(self, other):              # Change/ define the behaviour of "+" operator 
        return (self.x + other.x, self.y + other.y) # Here self.x = obj(p_add_1) and other.x = obj(p_add_2)
```

```python
p_add_1 = NewAdd(100,200)      # p_add_1 and p_add_2 will work becuase it is object of the NewAdd class and above in the add function we have mentioned self.x + other.x and same for y
p_add_2 = NewAdd(100,200)
```

```python
p_add_1 + p_add_2   # __add__ will only run when we have two objects 
```

```python
# 5. eq

class NewAdd:
    def __init__(self, x, y):
        self.x = x 
        self.y = y
    
    def __repr__(self):
        return f"NewAdd({self.x}, {self.y})"
    
    def __eq__(self, other):              
        return self.x == other.x and self.y == other.y
```

```python
p_add_1 = NewAdd(100,200)      
p_add_1 = NewAdd(100,200)      
```

```python
p_add_1 == p_add_2
```

```python
p_add_1 = NewAdd(101,200)      
p_add_1 = NewAdd(100,200)  
```

```python
p_add_1 == p_add_2
```

# Getter and Setter Methods 

```python
class Person:
    def __init__(self, name):
        self.__name = name # Private attribute or variable 

    def get_name(self):
        return self.__name 
    
    def __str__(self):
        return f"Person name is {self.__name}"
    
    def set_name(self, new_name):
        if not new_name:
            raise ValueError("Name cannot be empty")
        self.__name = new_name 

    def delete_name(self):
        del self.__name             # This is we are deleteing from the object
```

```python
p_gsd = Person("Prathamesh")
```

# Property Decorator 

```python
class Person:
    def __init__(self, name):
        self.__name = name # Private attribute or variable
        self.__name1 = name
        self.__name2 = name

    @property         # This is a docorater 
    def name(self):
        return self.__name 
    
    def __str__(self):
        return f"Person name is {self.__name}"
    
    @name.setter                  # Here we are linking decorator with the setter property using the name function 
    def name(self, new_name):
        if not new_name:
            raise ValueError("Name cannot be empty")
        self.__name = new_name 

    @name.deleter  # Here we are linking with the decoator 
    def name(self):
        del self.__name             # This is we are deleteing from the object
```

```python
p_property = Person("Monal")
```
