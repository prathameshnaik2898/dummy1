# Time Formatting 
### time.strftime(format, t)

### %M - minutes
### %m - month
### %d - date
### %Y - YYYY Year format
### %H - Hour
### %S - Seconds
### %y - YY year format

```python
print(time.strftime("%Y-%m-%d %H:%M:%S", time.localtime()))
```

```python
print(time.strftime("%Y-%m-%d", time.localtime()))
```
