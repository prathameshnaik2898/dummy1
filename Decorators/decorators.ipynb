# Inheritance: What if multiple parents have same method name and how to invoke the selected 


```python
class B:
    def greet(self):
        print("Hello from B")

class C:
    def greet(self):
        print("Hello from C ")

class A(B):               # This is B getting inherited into A 
    def greet(self):
        super().greet()      # Super will create a object of class B and it will run greet method present inside class B
        print("Hello from A")
```

```python
cls_obj_1 = A()
```

```python
cls_obj_1.greet()
```

```python
class B:
    def greet(self):
        print("Hello from B")

class C:
    def greet(self):
        print("Hello from C ")

class A(B,C):               # This class is narrating both B and C
    def greet(self):
        print("Hello from A")
```

# Decorator
### Decorator is the function that modifies the behaviour of another function and dosent change the code

```python
def my_decorator(func):
    def wrapper():
        print("Before the function runs.")
        func()
        print("After the Function runs.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
import time                     # STEP 0: Import time module for sleep() and time()

def research():
    # STEP 4: This is the original function we want to measure
    print("Start")
    time.sleep(5)               # Pauses execution for 5 seconds
    print("Complete")

def time_it(func):
    # STEP 1: time_it receives the function `research` as argument

    def wrapper():
        # STEP 3: wrapper is the function that replaces `research`
        # It adds extra behavior (timing) before and after research()

        t1 = time.time()        # STEP 5: Capture start time
        print(f"Time start is :{t1}")
        func()                  # STEP 6: Call the original function (research)
        t2 = time.time()        # STEP 7: Capture end time
        print(f"Time start is :{t2}")

        t = t2 - t1             # STEP 8: Calculate total execution time
        print(f"Time taken is: {t} seconds")

    return wrapper              # STEP 2: Return wrapper (NOT calling it)

# STEP 9: research is passed to time_it
# research becomes func
# a now points to wrapper
a = time_it(research)

# STEP 10: Calling a() actually runs wrapper()
a()
```

```python
# New Function 

def log_time_it(func):
    def wrapper(*args,**kwargs):   # args data type is tuple # **kwargs - to have d = 4 running because it is a keyword argument 
        t1 = time.time()
        print(f"Time start is :{t1}")
        func(*args,**kwargs)
        t2 = time.time()
        print(f"Time start is :{t2}")
        t = t2 - t1
        print(f"Time taken is: {t} seconds")
    return wrapper
```

```python
@log_time_it

def add(a,b,c,d):
    print(a + b + c)
    print(a + b + c)
    time.sleep(2)
    print(a + b + c)
    print(a + b + c)
```

```python
# Example to show when to use class method

class Memory:
    total_memory = 128

    @classmethod               # We need to add classMethod as a decorater to make use of it and to start a classMethod
    def update_memory(cls, m):   # First method in clasMethod is having cls as the first argument 
        cls.total_memory -= m      # cls is the reference to class Memory and we are using total_memory 
```

```python
Memory.total_memory
```

```python
Memory.update_memory(100)
```

# Static Method 

### These are part of the class but are not linked to the class and also dosent require self or cls

```python
class Calcualtor:

    @staticmethod
    def add(a,b):
        print(a + b)

    @staticmethod
    def multiply(a,b):
        print(a*b)

    def all(self, a,b):
        print(a + b)
        print("doing something")
        print(a * b)
```

```python
Calcualtor.add(1,2)
```

```python
c = Calcualtor()
```

```python
c.all(1,2)
```

```python
c.add(1,2)
```

```python
# These methods dosent depend on self or cls variable they are present on object as well as class 
class Calcualtor:

    @staticmethod
    def add(a,b):
        print(a + b)

    @staticmethod
    def multiply(a,b):
        print(a*b)

    def all(self, a,b):
        print(Calcualtor.add(a,b))
        print("doing something")
        print(Calcualtor.multiply(a,b))
```

```python
Calcualtor.add(1,2)
```

```python
c = Calcualtor()
```

```python
c.all(1,2)
```
