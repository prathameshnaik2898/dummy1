```python
import requests
import time
import threading 
```

```python
def download_file(process_name, url, file_path ):      # file_path → where to save the file on your system
    try:
        print(f"Download process name Started :{process_name}")
        response = requests.get(url)        # Sends a GET request to the URL, Downloads the content into response,response.content → raw bytes of file
        
        with open(file_path, "wb") as file:         # w is write mode(over writes if file exists) and b is binary mode
            for chunk in response.iter_content(chunk_size = 8192): # Reads the download file piece by piece,Each chunk is 8192 bytes(8KB)
                if chunk:
                    file.write(chunk)
        print("File downloaded Successfully")
    except Exception as e:
        print(f"Error downloading file: {e}")
    print(f"Process name completed: {process_name}")
```

```python
url = "https://raw.githubusercontent.com/Monalsingh/VideoBroadcaster/refs/heads/main/static/default-office-animated.png"
```

```python
t1_wd = threading.Thread(target = download_file, args = ("abc_1",url, "a.png"), daemon = True)  # Daemon is used to close the thread once the thread is completed
```

```python
t1_wd.start()
t1_wd.join()

print("Process Completed")
```

# CLose Manually Thread 

```python
stop_signal = threading.Event()  # by deafult is false

def watch_file(file_path):
    count = 0
    print(f"Watching{file_path}")
    while not stop_signal.is_set():         # While Not True <--- This will exit the thread but not close the thread. We now have the complete control 
        print("Watching file.....")
        time.sleep(2)
        count +=1
    # print("Watch exitinf cleanly...")

t1_wd = threading.Thread(target = watch_file, args = ("a.txt",))
t1_wd.start()

stop_signal.set() # Event Value to be True <--- This value works on thread level . We would have to make this value on and off

t1_wd.join()
time.sleep(5)

print("Main Program exited cleanly...")
```

# Manual Locking Using Dictionary

```python
stop_event = {"thread_1": False, "thread_2": False, "thread_3": False}
stop_event_lock = threading.Lock()

def watch_file(thread_id, file_path, thread_name):
    count = 0
    print(f"Watching{file_path}")
    while True:
        with stop_event_lock:
            if stop_event[thread_name]:         # While Not True <--- This will exit the thread but not close the thread. We now have the complete control 
                break
        print("Watching file.....")
        time.sleep(2)
    print(f"Watch thread id: {thread_id} exiting cleanly...")

t1_wd = threading.Thread(target = watch_file, args = (1, " a .txt",stop_signal_1,))
t2_wd = threading.Thread(target = watch_file, args = (2, " b .txt",stop_signal_2,))
t3_wd = threading.Thread(target = watch_file, args = (3, " c .txt",stop_signal_3,))

t1_wd.start()
t2_wd.start()
t3_wd.start()

time.sleep(5)
with stop_event_lock:
    stop_event["thread_1"] # Event Value to be True <--- This value works on thread level . We would have to make this value on and off

time.sleep(5)
with stop_event_lock:
    stop_event["thread_2"]

time.sleep(5)
with stop_event_lock:
    stop_event["thread_3"]

t1_wd.join()
t2_wd.join()
t3_wd.join()


time.sleep(5)

print("Main Program exited cleanly...")
```

# ThreadPoolExecutor 
### For every thread we have to write start and join, what if we dont want to do so 

```python
from concurrent.futures import ThreadPoolExecutor
```

```python
# Start
# join()

a = [(1, "a.txt","thread_1",),(2, "b.txt", "thread_2",),(3, "c.txt", "thread_3",),(4, "d.txt", "thread_4",)]
def watch_file(input_tuple):
    thread_id, file_path, thread_name = input_tuple
    count = 0
    print(f"Watching{file_path}")
    while count < 5:
        print("Watching file.....")
        time.sleep(2)
        count += 1
    print(f"Watch thread id: {thread_id} exiting cleanly...")
    return f"Hello {thread_id}"


with ThreadPoolExecutor(max_workers = 3) as executor:         # Total no of threads we defined, 
    results = list(executor.map(watch_file, a))

print("Thread Executed Successfully")

# watch_file,  (1, "a.txt","thread_1",)
# watch_file,  (2, "a.txt","thread_2",)
# watch_file,  (3, "a.txt","thread_3",)
# watch_file,  (4, "a.txt","thread_4",)
```
